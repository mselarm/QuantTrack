### [3.1.1. Conceptos básicos](#Índice)

- ### Clases

    - Las **clases** son el mecanismo que posee Java para describir principalmente:
        - Nuevas estructuras de datos, denominadas **propiedades** o **atributos**
        - Las operaciones o funciones que se pueden realizar con estas estructuras de datos y se denominan **métodos**

    - Tanto a los atributos como a los métodos se les denominan genéricamente **miembros** de la clase

    - El nombre o identificativo de la clase suele ir en **Mayúscula** la primera letra y ser un **nombre**. También se suele emplear el formato _(C)amelCase_



- ### Objetos

    - **Objeto**: representación en memoria que contiene todos los miembros de la clase, es decir atributos y métodos.
        - Un objeto es una **instancia** de una clase. 
        - Se crea con el operador `new` (ejecutándose -opcionalmente- una función interna de la clase denominada _constructor_ que puede admitir parámetros)
            ```java
            // Creación de un objeto
            MiClase objeto = new MiClase(...);
            ```
        - Acceso a **miembros** del *objeto*: 
            - Acceso a atributos (lectura o escritura, si es posible): `objeto.atributo`
            - Acceso a métodos (ejecución): `objeto.metodo(...)`
    - Los objetos en Java se manipulan con referencias:
        - Variable que simbólicamente _apunta_ a un objeto

- ### Comparación clases/tipos - objetos/variables
    - Declaraciones:
        - De una variable `x` de tipo `int`:
            ```java
            int x;
            ```
        - De un objeto `obj` de clase `MiClase`:
            ``` java
            MiClase obj;
            ```

    - Definiciones e inicializaciones de variables u objetos ya declarados:
        - De la variable:
            ```java
            x = 7;
            ```
            
        - Del objeto:
            ```java
            obj = new MiClase(<ARGUMENTOS>);
            ```

    - Declaracion y definición-inicialización:
        - De una variable `x` de tipo `int`:
            ```java
            int x = 7;
            ```
        - De un objeto `obj` de clase `MiClase`:
            ``` java
            MiClase obj = new MiClase(<ARGUMENTOS>);
            ```
    

- Por tanto:
    - Clase es a tipo...
    - ... lo que objeto es a variable

#### Descripción simbólica de un objeto de cierta clase. 


- Una **clase es**: un molde, una matriz, una plantilla
    - ... a partir de la cual, se crean **objetos** (mediante el operador `new`)

- ### Sintaxis simplificada de una clase

````java
class FuncionAfin {
    
    // Atributos (variables de la clase)
    double a;
    double b;
    
    // Constructor
    FuncionAfin(double valor_a, double valor_b) {
       a = valor_a;
       b = valor_b;
    }
    
    // Método de cálculo de la función en un punto x
    double calcular (double x) {
        return a * x + b;
    }
    
    // Método de cálculo de la función en una serie de puntos x: sobrecarga
    double[] calcular (double [] X) {
        int i;                                       // variable local
        double[] resultados = new double[X.length];  // array local
        for (i = 0; i < X.length; i++) {
            resultados[i] = calcular(X[i]);
        }
        return resultados;
    }
    
}
````

````java
FuncionAfin f = new FuncionAfin(2, 1);

System.out.println("f(5) = " + f.calcular(5));

double [] vector = {1., 2., 3., 4., 5.};
double [] resultados = f.calcular(vector);
for (int i = 0; i < vector.length; i++) {
   System.out.printf("f(%.1f) = %.1f\n", vector[i], resultados[i]);
}
````

### [3.1.2. Declaración e inicialización de atributos y variables](#Índice)

- Recordemos que dentro de una clase u objeto podemos encontrarnos _miembros_ como:
    - Atributos o propiedades
    - Métodos o funciones
    - Incluso otras clases
    - ...
- Matiz entre atributo y variable (local) estrictamente hablando:
    - **Atributo**: variable que pertenece a un **objeto** (o a una **clase**). 
        - También se denominan variables miembro
        - También se denominan variables de instancia cuando tienen presencia al instanciar una clase en un objeto (hay atributos que no tienen exactamente esta característica: se estudiarán más adelante)
    - **Variable**: variable que pertenece a un **método** (abusando del lenguaje, a los atributos también se les denomina variables del objeto). 
        - Cualquier _ítem_ que pertenezca al contexto de un método tendrá el calificativo de **local** (al método)
        - Aunque sea redundante hablaremos de variables _locales_ a aquellas que han sido definidas dentro de un método, 
- A partir de ahora se hará esa distinción semántica.
- Un atributo se declara sintácticamente igual que lo haría una variable local


#### Inicialización
- Atributos de un objeto:
    - Inicializados implícita y automáticamente
        - Variables de tipos primitivos:
            - Variables numéricas/caracteres: a `0`
            - Variables booleanas: `false`
        - Objetos: `null`
- Variables/objetos **locales** (a métodos) -los arrays tienen una consideración especial-:
    - **no** son inicializadas implícitamente
    - Si su primer acceso es de lectura: error en tiempo de compilación


El término **local** siempre hace referencia a que pertenece a un **método**

Ejemplos de inicialización implícita de variables primitivas y objetos como atributos y como variables locales:

````java
class Inicializacion {
   
    class OtraClase {
        // Descripción de la clase: atributos y métodos
    }

    // Se declaran con el modificador static para que puedan ser accedidas desde un metodo estático: main
    // Este detalle se estudiará posteriormente
    static int i;
    static float x;
    static char a;
    static boolean z;
    static String linea;
    static OtraClase objeto;

    public static void main(String[] args) {
    
        System.out.println("i: " + i );
        System.out.println("x: " + x );
        System.out.println("a: " + a );
        System.out.println("(short)a: " + (short)a );
        System.out.println("z: " + z );
        System.out.println("linea: " + linea );
        System.out.println("objeto: " + objeto );
              
        int local_i;
        String hola;
        // Error: uso sin inicializar!!!
        // System.out.println("local_i: " + local_i);
        // Los objetos locales no se inicializan a nada tampoco
        // if (hola == null) System.out.println("hola inicializado a null");
        
    } // main
    
} // Inicialización

Inicializacion.main(null);
````

Inicialización implícita de **arrays** de cierto tipo como **atributos** y como arrays **locales**:
- En ambos casos: a nivel escalar, como si fueran variables primitivas miembro o atributos
    - Tipo numérico/caracteres: a `0`
    - Tipo booleano: `false`
    - Objetos: `null`
    
Ejemplos:    

````java
class Inicializacion_arrays {
   
    // Se declaran con el modificador static para que puedan ser accedidas desde un metodo estático: main
    // Este detalle se estudiará posteriormente
    static int[]     vector_int    = new int[2];
    static double[]  vector_double = new double[2];
    static boolean[] vector_bool   = new boolean[2];

    public static void main(String[] args) {
    
        // array atributo o miembro
        System.out.println("vector_int[0]:          " + vector_int[0]       + ",     vector_int[1]:           " + vector_int[1]);
        
        // array local
        int[] vector_int_local = new int[2];
        System.out.println("vector_int_local[0]:    " + vector_int_local[0] + ",     vector_int_local[1]:     " + vector_int_local[1]);

        // array atributo o miembro
        System.out.println("vector_double[0]:       " + vector_double[0]       + ",   vector_double[1]:       " + vector_double[1]);
        
        // array local
        double[] vector_double_local = new double[2];
        System.out.println("vector_double_local[0]: " + vector_double_local[0] + ",   vector_double_local[1]: " + vector_double_local[1]);

        // array atributo o miembro
        System.out.println("vector_bool[0]:         " + vector_bool[0]       + ", vector_bool[1]:         " + vector_bool[1]);

        // array local        
        boolean[] vector_bool_local = new boolean[2];
        System.out.println("vector_bool_local[0]:   " + vector_bool_local[0] + ", vector_bool_local[1]:   " + vector_bool_local[1]);
        
        
    } // main
    
} // Inicialización

Inicializacion_arrays.main(null);
````

#### Inicialización explícita de arrays

Inicialización de arrays y arrays anónimos:

````java
void metodo(String[][] a) {
    System.out.println("a.length = " +  a.length);
    System.out.println("a[0].length = " +  a[0].length);
    System.out.println();
}


// Inicialización componente a componente: DIMENSIONADO (reserva o definición) junto la declaración
String[][] nombres = new String[2][2];
// O bien por separado, declaración y reserva:
//   String[][] nombres; 
//   nombres = new String[2][2];
nombres[0][0] = "string 0-0";
nombres[0][1] = "string 0-1";
nombres[1][0] = "string 1-0";
nombres[1][1] = "string 1-1";
metodo(nombres);


// Inicialización literal de un array: DIMENSIONADO o definición automática en la propia inicialización 
String[][] nombres_2 = { {"Hola", "Adios", "Hasta luego"}, 
                         {"Mañana", "Hoy", "Ayer"}, 
                         {"Derecha", "Izquierda", "Arriba"} 
                       };
metodo(nombres_2);


// El argumento es una matriz (array) de strings ANONIMA (DIMENSIONADO o definición automática en la creación/inicialización)
metodo( new String [ ][ ] { {"Pepe", "Manolo", "Felipe"}, 
                            {"Gutiérrez", "González", "López"}
                          }  
      );

````

### [3.2.2. Constructor de un objeto](#Índice)

#### Versión 2: con constructor

- Hay propiedades que definen ya algunas funcionalidades de un objeto:
    - Su ajuste inicial (y probablemente definitivo, podrían ser perfectamente constantes tras su inicialización) hace que el objeto sea al menos parcialmente operativo. En nuestro ejemplo:
        - `nserie`
        - `largo`, `ancho`, `alto`
        - `peso`
        
        La propiedad o atributo `volumenContenido` **no** será previsiblemente una constante, ni tampoco el `precio` pues ambos, por motivos distintos, pueden variar a lo largo de la vida del objeto `deposito`.
        
    - Dicho ajuste inicial puede darse en una _función_ especial denominada **constructor**
        - Se invoca de forma automática una sola vez por objeto creado: cuando se crea el objeto mediante el operador `new`
        - Tiene el **mismo nombre** que la clase
        - Es una especie de método más, pero **no** tiene tipo de retorno ni es invocable convencionalmente, y mucho menos, fuera del entorno de la clase
        - De momento, se le añade el modificador `public` denotando que _todo el mundo_, es decir, desde _cualquier otra clase_ se puede emplear el constructor, luego se puede crear o construir un objeto de dicha clase.
        - Los atributos con modificador `final` (constantes)  **deben** ajustar su valor en:
            - bien su definición:
                ```java
                final int pi = 3.1415;
                ```
            - o bien en el constructor de la clase a la que pertenecen:
                ```java
                final int constante; // pendiente de ajustar el valor definitivo en un constructor; 
                                     // si no se ajusta en un constructor: error
                ```
            En nuestro caso, los atributos que considerarmos `final` serán:
            - `nserie`
            - `largo`, `ancho`, `alto`
            - `peso`

### [3.2.3. Parámetros (de constructores y métodos) y atributos](#Índice)

#### Versión 3: ambigüedad en nombres de variables
- Se pueden emplear el mismo nombre para una variable **parámetro** que para un **atributo**, pero con consecuencias... 
    - Si la intención es `atributo`<- `parámetro` dentro del constructor (o de cualquier método de la clase), se puede deshacer la ambigüedad o _sombra_ del parámetro sobre el atributo:
        - Utilizando el prefijo `this` en atributo
            ```java
            public class Deposito {
    
                final String nserie; // atributo del depósito
    
                // constructor
                public Deposito(String nserie, ...) {
                    this.nserie = nserie;
                    ...
                }
                
                ...
            }
            ``` 
        - `this` denota el objeto que se está formando, por lo que `this.nserie` es sin ningún tipo de ambigüedad el atributo del depósito; `nserie` es el parámetro del constructor (el cual tiene prevalencia sobre el atributo, ya que está declarado explícitamente en el propio constructor como parámetro de tipo `String` en este caso).

### [3.2.4. Control de lectura/escritura: modificador `private` y métodos _getter_ y _setter_ ](#Índice)

#### Version 4: control de acceso
- Tras su inicialización, conviene que algunos atributos (algunos miembros, en general, incluyendo por tanto a los métodos) no sean accesibles desde **fuera** de la clase: esto puede llevarse a cabo cambiando el modificador de `public` a `private`. El concepto de gestinar el acceso a atributos de un objeto desde el exterior se denomina _encapsulamiento_.
    - Algunos de los atributos, ni siquiera interesa que se modifiquen tras su inicialización: puede conseguirse con la adición de modificador `final` en la declaración de los mismos (ya estudiado previamente).
    - En cualquier caso, los miembros con modificador `private` **pueden seguir siendo accedidos desde _dentro_** de la clase tanto para lectura como para escritura (solo lectura para aquellos con modificador `final`).
    - Los miembros con modificador `private` **no** puede ser accedidos _desde fuera de la clase_  ni para lectura ni para escritura.
    - Al haber prohibido el acceso por diseño (uso del modificador `private`), si de todas formas se desea acceder a los atributos privados _desde fuera de la clase_, todavía se puede hacer si dentro de la clase diseñamos _ciertos_:
        - **Métodos** _públicos_ (funciones con modificador `public`)      
        - Los métodos (públicos) con estos roles de acceso a variables privadas se suelen denominar `getter` y `setter`:
        
            - Si el método lee atributo: método _getter_
                ```java
                public <TIPO_ATRIBUTO> get<ATRIBUTO>(){ return <ATRIBUTO>; }
                ```
            - Si el método escribe atributo: método _setter_ 
                ```java
                public void set<ATRIBUTO>(<TIPO_ATRIBUTO> <ATRIBUTO>){ this.<ATRIBUTO> = <ATRIBUTO>; }
                ```
            
        - El IDE puede ayudar a ello: botón derecho sobre fichero clase -> Generar -> escoger método getter/setter/ambos

### [3.2.5. Métodos y sus sobrecargas](#Índice)

#### Versión 5: nuevos métodos y sobrecargas

- Se propone diseñar un método público que devuelva la capacidad del depósito en litros: `largo x ancho x alto` denominado `public float getCapacidad()`

- Con la intención de tener un objeto más y mejor controlado, se añaden varios métodos adicionales:
    - Para llenar el depósito con cierta cantidad de litros especificada en un parámetro: `public void llenar(float litros)`
        - Se incrementará el volumen del contenido del depósito en la cantidad indicada en el parámetro
        - Se debería llevar cierto control sobre un posible rebosamiento del depósito. Dos posibilidades:
            - Que al llenar no haya rebosamiento   
            - Que al llenar haya rebosamiento. En este caso se procederá de la siguiente manera: el volumen del contenido se ajustará al máximo posible (capacidad del deposito)
        
          Para indicar si hay rebosamiento o no se devolverá un booleano con tal circunstancia, luego se cambiará el tipo de retorno de `void`a `boolean`, es decir: `public boolean llenar(float litros)`

    - Consumo del depósito: `public void vaciar(float litros)`
        - Se reduce en `litros` el volumen del contenido del depósito
        - Se debería llevar cierto control sobre un posible vaciado del deposito. Dos posibilidades:
            - Que hayan suficientes litros: se reduce el volumen del contenido en la cantidad indicada
            - Que no hayan suficientes litros: el volumen del contenido se deja a cero
          
          Para indicar que han ocurrido alguna de las dos circunstancias anteriores, puede devolverse un booleano, luego al final se diseñará como `public boolean vaciar(float litros)`  
    
    - Para llenar el depósito _"hasta arriba"_: se empleará una _sobrecarga_ de `llenar` sin especificación de litros
        
        - Para nuestro ejemplo: `public float llenar()` se interpretará como llenar el depósito hasta arriba (podemos considerar que es una sobrecarga de `public boolean llenar(float litros)`). Parece razonable suponer que no tiene sentido plantearse la posibilidad de rebose, ya que se repostará con _lo justo_ para no rebosar. 
     
    - Para vaciar completamente el depósito se empleará una _sobrecarga_ de `vaciar` sin especificación de litros.
    
- Con estos métodos planteados, quizá no tenga sentido mantener `setVolumenContenido(litros)` ya que no tiene ningun tipo de control, luego, se eliminará.

### [3.2.6. Sobrecarga del constructor](#Índice)

#### Versión 6: sobrecarga del constructor
    
- Se va a _sobrecargar_ el **constructor** con la intención de generalizar algo más la creación del objeto. En nuestro caso vamos a omitir en el nuevo constructor el valor del volumen del contenido, provocando que dentro del objeto tome como valor el valor por defecto de un atributo numérico sin inicializar (es decir, cero; por lo tanto el volumen del contenido inicial será nulo al emplear esta sobrecarga del constructor). Luego las sobrecargas disponibles ahora serán:
    - La nueva sobrecarga: `public Deposito(int nserie, float largo, float ancho, float alto, float peso, float precio) {...}`

#### Asignación

#### Version 11: comparación de objetos


- La mecánica de comparación de objetos también es **muy relativa**: depende de nuestra aplicación.
    - Dos depósitos son iguales si ocupan a la misma posición en memoria en el ordenador de simulación
    - Dos depósitos son iguales si tienen el mismo número de serie (aunque sean dos objetos diferentes)...
    - Dos depósitos son iguales si tienen el mismo número de serie y la misma capacidad...
    - ...
    - Depende...
- Dada tanta casuística, siendo `a` y `b` dos objetos, la operación de comparación o igualdad entre ambos, `a==b` en Java hace referencia a la comparación de las direcciones de memoria donde están almacenados. Ejemplo:        
    ```java
    Clase x = new Clase(1);
    Clase y = x;
    Clase z = new Clase(2);
    Clase w = new Clase(1);
    
    boolean comparacion = (x == y);
    comparacion = (x == z);
    ...
    ```
        
    | `a` | `b` | `a == b`  | Comentario |  
    |:---:|:---:|:---------:|:-----------|  
    | `x` | `y` | `true`    | `x` e `y` apuntan a misma posición de memoria (puntero de inicio) |  
    | `x` | `z` | `false`   | `x` y `z` son objetos distintos ubicados en posiciones de memoria distintas |  
    | `x` | `w` | `false`   | `x` y `w` son objetos distintos ubicados en posiciones de memoria distintas (aunque idéntico contenido) |  

- Para la comparación de objetos, la operación `==` suele ser **insuficiente**. Java proporciona dos herramientas:
    - Método para comparar la igualdad de objetos, comparando atributos o propiedades de los dos objetos: `equals()`. Condiciones:
        - Si son el mismo objeto, se devuelve `true`
        - Si el objeto con el que se va a comparar es `null` o son de clases distintos, se devuelve `false`
        - Si todas las propiedades o atributos son _iguales_, entonces se devuelve `true`; en caso contrario, `false`
        - El parámetro de `equals` es de la clase `Object` para que sea una sobreescritura del método de su superclase (se verá posteriormente)
    - Método para comparar identidades: `hashCode()`
        - Un _hashCode_ es un entero que identifica a un objeto
        - Dos objetos que sean _iguales_ deben tener el mismo _hashCode_
        - Debe existir una extraordinariamente pequeña probabilidad de que dos objetos distintos tengan el mismo _hashCode_
        - Existe un método (`Objects.hash(<ARGUMENTOS_LONGITUD_VARIABLE>)`, método _estático_ de la clase `Objects` -que no `Object`-) que devuelve un `hashCode` calculado a partir de la lista de argumentos proporcionados 
        - Se utiliza en otras clases como `HashSet` y `HashMap`: todo funcionará bien si siendo dos objetos iguales según `equals`, también lo son según `hashCode()`

    
- Ejemplo:
    ```java

    class MiClase {

        int x;
        String y;

        ...

        @Override
        public boolean equals(Object obj) {
            if (this == obj) return true; // misma referencia
            if (obj == null || getClass() != obj.getClass()) return false;

            Clase elOtroObjeto = (Clase) obj;
            return x == elOtroObjeto.x && Objects.equals(y, elOtroObjeto.y);
        }

        // Método hashCode: consistente con equals
        @Override
        public int hashCode() {
            return Objects.hash(x, y);
        }

    }

    ```
- Estos métodos pueden ser generados automáticamente en el IDE:

## [3.3. Paquetes _(packages_)](#Índice)

<div id="package"></div>

### [3.3.1. Concepto de `package`](#Índice)

- Un paquete o *package* es un esquema de almacenamiento y de referencia de clases, tanto fuentes como compiladas en _byte-code_:
    - Agrupación de clases afines en un mismo paquete: equivalente a librería en otros lenguajes
    - Una clase perteneciente a un paquete puede usar con cierta *preferencia* otras clases o miembros de estas, definidas en el mismo paquete
- Los `package` también delimitan el espacio de nombres 
    - El nombre de una clase debe ser único en el `package` donde se define
    - Dos clases con el mismo nombre, pero distinto `package` pueden coexistir
- Los `package` **no** se puede *emular* en un Jupyter Notebook    

- Una clase se declara perteneciente a un *paquete* con la cláusula `package`
    ```java
    package <NOMBRE_PAQUETE>;
    ```

- Debe ser la **primera** sentencia del fichero fuente.
- La clase declarada en ese archivo *pertenece* a ese `package`. Ejemplo:

    ```java
    package miPackage;    
    ...    
    class MiClase {
        ...
    }
    ```
    
    La clase `MiClase` pertenece al paquete `miPackage`

- Si no hay cláusula `package`, la clase pertenece a un paquete por defecto sin nombre. 
    - Suele ser **conveniente** poner nombre de paquete.

<div id="cualificado"></div>

### [3.3.2. Nombre cualificado de la clase](#Índice)

El nombre **cualificado** de la clase se forma con el paquete como prefijo: `<NOMBRE_PAQUETE>.<CLASE>`

Ejemplo (fichero `MiClase.java`):

```java
package miPackage;     %%Es el directorio

...    
class MiClase {
    ...
}
```

El nombre **cualificado** de la clase será: `miPackage.MiClase`.

La referencia a una clase definida en un paquete será **siempre** por su **nombre cualificado**.


Hay que distinguir:
- Nombre de clase: `MiClase`
- Nombre cualificado de la clase: `miPackage.MiClase`
- Fichero fuente de la clase: `MiClase.java`
- Fichero en _byte code_ de la clase: `MiClase.class`
    - Ubicación _relativa_ del fichero de la clase en el sistema de ficheros del ordenador: `miPackage/MiClase.class`
  

<div id="esquema"></div>

### [3.3.3. Esquema de nombres de paquetes y ubicación](#Índice)

- La sintaxis del nombre del paquete suele ser una secuencia de nombres en minúsculas separados por punto. Por ejemplo:
```java
package p1.p2.p3;
...
class Clase {
    ...
}     
```

No existe el concepto de _subpaquete_: se suele interpretar que 
   - `p1.p2` puede considerarse un *paquete* que contiene clases, sin embargo, `p2` no se interpreta como *subpaquete* de `p1`, ni `p1.p2.p3` se considera _subpaquete_ de `p1.p2`: `p1.p2.p3` es un paquete _independiente_ del paquete `p1.p2`.
   

En ocasiones, los nombres separados por puntos denotan algún tipo de estructura _organizativa_ interna de paquetes.

- El nombre del paquete debería ser **único**. Convenio: nombre del dominio al revés. Ejemplo:
```
es.upv.etsit.asignatura.funcionalidad
```
- **No** debería emplearse `java` o `javax` como *prefijo* del paquete, por ser prefijos estándar ya utilizados.


#### Ubicación

- En función de cómo se realice la compilación, el propio nombre del paquete indica dónde **estará** ubicado el fichero de la clase tras compilar el fichero fuente, respecto a cierto directorio de referencia `<REF>`:

    ```java
    // Fichero Clase.java
    package p1.p2.p3;           <REF>
    class Clase {                 └── p1 
      ...                  ->          └── p2 
    }                                       └── p3 
                                                 └── Clase.class
    ```
   
    o bien:
    ```bash
    <REF>/p1/p2/p3/Clase.class
    ```

    o bien:
    ```bash
    <REF>\p1\p2\p3\Clase.class
    ```
    en función del sistema operativo.

- El directorio de referencia `<REF>` puede fijarse:
    - Si se compila en línea de comandos, con la opción `-d`: 
    `javac -d <REF> ...`
    - Si se compila con el IDE, dependerá del _Build system_ (en la etapa de configuración del proyectdo: IntelliJ, Maven, Gradle, ...).
        - Con el esquema `IntelliJ` `<REF>` = `out\production\<NOMBRE_DEL_PROYECTO>`

- Es muy conveniente que los ficheros fuente `*.java` respeten también esa estructura de almacenamiento. El IDE suele ayudar en ello.

### [3.3.4. Cláusula `import`](#Índice)

- Permite indicar al compilador el uso de una clase concreta o de todas de las de un paquete.
- **Importante:**
    - `import` realmente **no** importa paquetes, importa clases **cualificadas**
- Sintaxis:
    - `import <PAQUETE>.<CLASE>`: importa *exclusivamente* la clase `<CLASE>` del paquete `<PAQUETE>`
    - `import <PAQUETE>.*`: importa **todas** las clases del paquete `<PAQUETE>`
- Los paquetes **no** se pueden referenciar de forma relativa a otros paquetes. Siempre hay que hacerlo de forma absoluta.    

- Ejemplo: 

    - Fichero `Clase1.java`
    ```java
    package p1.p2.p3;
    public class Clase1 {
        ...
    }
    ```        
    - Fichero `Test.java`:
    ```java
    package paquete1.paquete2;
    import p1.p2.p3.Clase1;
    // Alternativa:
    // import p1.p2.p3.*; // Importa TODAS las clases del paquete p1.p2.p3: se usen o no
    public class Test {
        Clase1 c1 = new Clase1();
        ...
    }
    ```

- Se debe respetar el orden indicado de las instrucciones `package` e `import`: 
    - Primeramente, el nombre del paquete al que pertenecerá la clase, mediante la sentencia `package`
    - Indicación de todas las clases a importar, mediante  sentencias `import`
    
    
    
    
- Si la clase a utilizar pertenece al mismo paquete, entonces no hace falta llevar a cabo la *importación*. 

    En el siguiente ejemplo, tanto la clase `Test` como `Clase2` pertenecen al mismo paquete `paquete1.paquete2` por lo que no es necesario la sentencia `import paquete1.paquete2.Clase2` en `Test.java`.
    - Fichero `Clase2.java`: pertenece a paquete `paquete1.paquete2`
        ```java
        package paquete1.paquete2;
    
        class Clase2 {
            ...
        }
        ```
    - Fichero `Test.java`: también pertenece a paquete `paquete1.paquete2` luego puede acceder directamente a `Clase2`
        ```java
        package paquete1.paquete2;
    
        class Test {
            Clase2 c2 = new Clase2();
            ...
        }
        ```
    
> Estas posibilidades de acceso pueden _modularse_ mediante modificadores de acceso que se estudiarán posteriormente.    


- Los paquetes no se pueden referenciar de forma relativa:
    
    - Fichero `MiClase.java`
        ```java
        package p1.p2.p3;

        public class MiClase {
            ...
        }
        ```        
    - Fichero `MiTest.java`:
        ```java
        package p1.p2;

        import p3.MiClase; // Error. Debería ser de forma absoluta: import p1.p2.p3.MiClase;

        public class MiTest {
            MiClase c = new MiClase();
            ...
        }
        ```

### [3.4.1. Comentarios de documentación](#Índice)

Añadiendo los comentarios con cierta estructura, mediante la aplicación `javadoc` contenida en el JDK se puede generar documentación de la clase en formato HTML. La estructura de un comentario de documentación **debe** ser la siguiente:

```java
/**
 * Comentario descriptivo
 */
```

Obsérvese cómo el arranque de comentario es con `\**`  y no con `\*`.

Internamente se puede añadir información para posteriormente generar la documentación con las siguientes consideraciones:

```java
/**
 * Descripción general de la clase o método.
 *
 * @param       nombreDescripción del parámetro
 * @return      Descripción del valor de retorno
 * @throws      TipoDeExcepcion Descripción de la excepción lanzada
 * @see         OtraClase o método relacionado
 * @author      Autor del código
 * @version     Versión del código
 * @deprecated  Marca un método o clase como obsoleto
 * ...
 */
```

#### Extracción de la documentación: `javadoc`

- En **línea de comandos**: ejemplo de ejecución de javadoc (véase `javadoc -help`) 
    - `-d`: directorio destino de toda la documentación HTML: `-d doc`
    - `-private`: se documenta todo (privado, publico, paquete, protegido, . . . ) 
    - `-charset` a especificar en HTML: `-charset utf8`
    - Especificación de fuentes a documentar:
        - Indicando `-sourcepath <DIRECTORIO>`
    - Al final, indicación de directorios de los que extraer la documentación        
- La estructura del proyecto se supone que es la que aparece a continuación y la intención es crear automáticamente el directorio `doc` y albergar ahí toda la documentación (el punto de entrada será el fichero `index.html`:  
    ```text
    <REF>
       ├── src
       │   └── espacioVectorial
       │       ├── coordenadas
       │       │   └── Coordenada2D.java
       │       └── geometria
       │           └── GeometriaEuclidea.java
       │
       ├── doc ### será creado automáticamente por javadoc
       │   ├── ...
       .   ├── index.html  ### punto de entrada de la documentación
       .   └── ... 
       .
    ```
    y suponiendo que el directorio de trabajo en el que estamos es `<REF>`

    ```bash
    $ javadoc -charset utf8 -d doc -private -sourcepath src espacioVectorial
    
    javadoc -d doc -version -author -private -sourcepath src -subpackages espacioVectorial

    
    ```

```java
package espacioVectorial.geometria;

/**
 * En esta interfaz se define el método que a implementar para poder establecer una geometría euclídea.
 *
 * La geometría euclídea se fundamenta en gran parte en el concepto de distancia entre dos puntos, definida
 * mediante el postulado de las rectas. La distancia determina la longitud del segmento rectilíneo más corto que
 * une dos puntos.
 *
 * @version 1.0
 * @author Paco
 */

public interface GeometriaEuclidea {

    /**
     * Distancia entre dos puntos o coordenadas
     *
     * @param b Objeto que represente un punto en el espacio euclídeo
     * @return distancia euclidea entre dos puntos o coordenadas del espacio euclídeo en cuestión
     */
    double distancia(Object b);
    // Se indica Object como la clase del parámetro para que sea válido cualquier tipo de vector

}
```

Fichero `Coordenada2D.java`:

```java
package espacioVectorial.coordenadas;

import espacioVectorial.geometria.GeometriaEuclidea;

import java.util.Objects;

/**
 * En esta clase se implementan todos los métodos para trabajar con coordenadas
 * en 2D.
 *
 * Las coordenadas están dadas en un espacio euclideo, luego se está obligado a
 * implementar todos los métodos necesarios para cumplir con los postulados
 * de la geometría euclídea.
 *
 * @version 1.1
 * @author Paco
 */

public class Coordenada2D implements GeometriaEuclidea {

    /** Coordenada x */
    private double x;
    /** Coordenada y */
    private double y;


    /**
     * Constructor de coordenada 2D.
     *
     * @param x coordenada {@link #x}
     * @param y coordenada {@link #y}
     */
    public Coordenada2D(double x, double y) {
        this.x = x;
        this.y = y;
    }


    /**
     * Constructor de coordenada 2D con coordenda y = 0.
     *
     * @param x coordenada {@link #x}
     */
    public Coordenada2D(double x) {
        this(x, 0.);
    }


    /**
     * Constructor de coordenada 2D con coordendas (x,y) = (0.,0.)
     *
     */
    public Coordenada2D() {
        this(0.);
    }


    /**
     * Constructor de copia.
     *
     * @param c coordenada2D a clonar
     */
    public Coordenada2D(Coordenada2D c) {
        this(c.x, c.y);
    }


    /**
     * Getter de la coordenada {@link #x}.
     *
     * @return coordenada {@link #x}
     */
    public double getX() {
        return x;
    }


    /**
     * Getter de la coordenada {@link #y}.
     *
     * @return coordenada {@link #y}
     */
    public double getY() {
        return y;
    }


    /**
     * Suma de cuadrados de las coordenadas
     *
     * @return cuadrado de coordenada x más cuadrado de coordenada y
     */
    protected double sumaDeCuadrados() {
        return x * x + y * y;
    }


    /**
     * Distancia entre dos puntos o coordenadas
     *
     * @param b coordenada o punto desde la que calcular la distancia
     * @return distancia euclidea entre dos puntos o coordenadas
     *
     */
    public double distancia(Object b) {
        // Aunque los atributos x e y son privados y por tanto no accesibles
        // en este contexto lo son (b.x y b.y) ya que estamos en la misma clase
        Coordenada2D bb = (Coordenada2D) b;
        return Math.sqrt(new Coordenada2D(x - bb.x,y - bb.y).sumaDeCuadrados());
    }


    /**
     * Compara dos coordenadas
     *
     * @param o la otra coordeanda
     * @return true si son iguales las componentes; false en caso contrario
     *
     */
    @Override
    public boolean equals(Object o) {
        // El parámetro se pone de la clase Object y no de la clase Coordenada2D
        // para que sea propiamente la sobrescritura del método equals de la superclase
        // es decir, para que tenga la misma firma
        if (o == null || getClass() != o.getClass()) return false;
        Coordenada2D that = (Coordenada2D) o;
        return Double.compare(getX(), that.getX()) == 0 && Double.compare(getY(), that.getY()) == 0;
    }

    /**
     * Método hashCode
     *
     * @return hashCode conjunto de las coordeandas x e y
     *
     */
    @Override
    public int hashCode() {
        return Objects.hash(getX(), getY());
    }

    /**
     * Método toString().
     *
     * @return ({@link #x}, {@link #y})
     */
    public String toString() {
        return "(" + x + ", " + y + ")";
    }

} // Coordenada2D

```
