#### Declaración de una variable primitiva o de tipo primitivo

```java
[<MODIFICADOR>] <TIPO> <NOMBRE_VARIABLE> [ = <VALOR_INICIAL> ] ;
```

donde `<MODIFICADOR>` es un _modificador_ de la variable (se verá posteriormente) y `<TIPO>` es el nombre del tipo primitivo al que pertenece la variable.

#### Declaración de un objeto

```java
[<MODIFICADOR>] <CLASE> <NOMBRE_OBJETO> [ = new <CLASE>(<PARAMETROS>) ] ;
```

donde `<MODIFICADOR>` es un _modificador_ de la clase y `<CLASE>` es el nombre de la clase a la que pertenece el objeto.



#### Notación:
- `<CONCEPTO>`: identifica el concepto que representa
- `[  ]`: indica opcionalidad (puede omitirse). Con el ejemplo anterior serían válidas las siguientes posibilidades:
    ```java
    <TIPO> <NOMBRE_VARIABLE> = <VALOR_INICIAL> ;
    ```
    O bien:
    ```
    <TIPO> <NOMBRE_VARIABLE> ;
    ```
    O bien:
    ```java
    <MODIFICADOR> <TIPO> <NOMBRE_VARIABLE> = <VALOR_INICIAL> ;
    ```
    O bien:
    ```
    <MODIFICADOR> <TIPO> <NOMBRE_VARIABLE> ;
    ```
- `|`: indica alternativa

### Nota aclaratoria sobre escritura en pantalla

- Se emplea la instrucción `System.out.println( <ARGUMENTO> )`, que será explicada posteriormente
- De momento, téngase en cuenta solo los siguientes detalles:
    - El argumento de `System.out.println( <ARGUMENTO> )` puede ser cualquier variable u objeto de cualquier tipo o clase
    - En los ejemplos se observará la estructura: `System.out.println("hola " +  <VARIABLE> )`
        - Un string _literal_ se muestra encerrado entre comillas dobles: `"hola "`
        - Como se explicará posteriormente, el operador `+` que aparece **no** realiza ninguna _suma_ algebraica cuando uno de los operandos es un _string_; lo que hace es **concatenar** el string literal `"hola "` con la conversión a string de `<VARIABLE>`, resultando al final un string más largo: este es el que se imprime en pantalla.
        - Así, por ejemplo, en el caso del _snippet_ anterior:
        ```java
        double x = 9.7;
        System.out.println("hola " + x);
        ```
        se imprime el string `hola ` y a continuación, concatenadamente, el valor numérico de x, es decir la salida sería `hola 9.7`
- Como anticipación adicional, recuérdese que todas la instrucciones de Java deben terminarse con el signo de punto y coma `;`.        
- El sangrado de las instrucciones es solo a nivel estético o visual: no aporta estructura de control o ejecución tal y como sí ocurre en **Python**.

### Boxing/unboxing entre variables primitivas y sus envoltorios o *wrappers*, ...

- **_Boxing_**: creación automática de un objeto a partir de una variable primitiva: **_tipo primitivo_ &#x2192; _objeto_**

- **_Unboxing_**: creación automática de una variable primitiva a partir de un objeto: **_tipo primitivo &#x2190; objeto_**

| Tipo primitivo | &#x21c6; | _Wrapper_ o envoltorio |
| :---           | :--: | :--- | 
|`byte` | &#x21c6; |`Byte`|
|`short` | &#x21c6; | `Short`|
|`int` | &#x21c6; |`Integer`|
|`long` | &#x21c6; |`Long`|
|`float` | &#x21c6; |`Float`|
|`double` | &#x21c6; |`Double`|
|`char` | &#x21c6; |`Character`|
|`boolean` | &#x21c6; |`Boolean`|

`````java
// Boxing
Double x;          // Objeto
double y = 10.0;   // variable primitiva

x = y; // Double <- double   (boxing o wrapping)
System.out.println("y = " + y + ", x = " + x);
`````

`````java
// Unboxing
Double x = new Double(0.5); // creación de un objeto de la clase Double: se estudiará posteriormente
double y = x; // double <- Double (unboxing o unwrapping)
System.out.println("y = " + y + ", x = " + x);
`````

### [2.1.2. _Casting_](#Índice)


- Conversión explícita de un tipo a otro
- Uso:
    - Indicación explícita de **asunción** de posible pérdida de precisión
    - Conversión indicando entre paréntesis el tipo destino
        - Ejemplo:
            - `<VARIABLE_TIPO_A> = (<TIPO_A>) <VARIBLE_TIPO_B>;`
            ```java
            double x = 1.23;
            int y;
            y = (int) x;
            ```
    - Si no hay pérdida de precisión en la asignación, el _casting_ no es necesario
        - Ejemplo:
            ```java
            double x;
            int y = 5;
            x = y;
            ```
    

`````java
double x = 1.234567892345;

int y;
y = x; // Pérdida potencial de precisión: ERROR

float z;
z = x; // Pérdida potencial de precisión: ERROR

System.out.println("x: " + x);
System.out.println("y: " + y);
System.out.println("z: " + z);
`````

#### Solución (y asunción de pérdida de precisión) con _casting_:

`````java
double x = 1.234567892345;

int y;
y = (int) x;   // Asunción de la pérdida potencial de precisión.

float z;
z = (float) x; // Asunción de la pérdida potencial de precisión.


System.out.println("x: " + x);
System.out.println("y: " + y);
System.out.println("z: " + z);
`````

`````java
int    y = 3;
double x = y;
float  z = y;
double w = z;

System.out.println("y: " + y);
System.out.println("x: " + x);

System.out.println("z: " + z);
System.out.println("w: " + w);

`````

### Concepto de **Widening** y **narrowing**:

- #### Widening

    - Conversión primitiva expansiva *(widening)* o *promoción* a tipo superior: **no hay pérdida de información** en la conversión
        - byte → short, int, long, float, double
        - short → int, long, float, double
        - char → int, long, float, double
        - int → long, float, double
        - long → float, double
        - float → double

- #### Narrowing

    - Conversión primitiva compresiva *(narrowing)*
        - short → byte, char
        - char → byte, short
        - int → byte, short, char
        - long → byte, short, char, int
        - float → byte, short, char, int, long
        - double → byte, short, char, int, long, float

        En este caso, se requiere *casting* por **pérdida potencial de precisión** en conversión
            - Si no se realiza el *casting*: **error** en tiempo de compilación




### Conversiones implícitas

- Aunque en una expresión aparezcan más de dos operandos, las operaciones aritméticas convencionales (suma, resta, multiplicación ...) son binarias: dos operandos
- A continuacion se exponen las conversiones implícitas entre el par de operandos en cuestión

#### Conversión en operaciones aritméticas donde puede o no haber mezcla de distintos tipos numéricos enteros
- **Widening** implícito: 
    - Siempre se _promocionan_ los tipos de menor precisión que `int` a tipos `int` (excepción cuando los operandos son _literales_ -se verá más adelante-):
    - Ejemplo: suma de un `byte` y un `short`: 
        - `<byte>` + `<short>`
        - `<int>` + `<int>`: _widening_ o promoción del `<byte>` a `<int>` y del `<short>` a `<int>`
        - Resultado: `<int>`
    - Si alguno de los dos operandos es un `<long>`: promoción del otro a `<long>`; resultado `<long>`        
- Si la **asignación** del resultado implica _narrowing_: si no hay _casting_, error en tiempo de compilación
    
#### Entre tipos reales: `float`  y `double`
- **Widening implícito**:
    - `float` promociona a `double`
- Si la **asignación** implica _narrowing_: si no hay _casting_, error de compilación
  
#### Entre tipos enteros y reales:
- ***Widening implícito**:
    - Todos los tipos enteros promocionan al tipo del otro operando real (`float` o `double`)
- Si la **asignación** implica _narrowing_: si no hay _casting_, error de compilación

#### Conversión en invocación a método:

- Cuando se invoca a un método, el tipo de los argumentos se _convierte_, si es necesario, al de los parámetros.
    - Si en la adaptación de argumentos a parámetros existe **_narrowing_** se produce un error de compilación.
    

### Ejemplos:

`````java
byte x = 1;
byte y = 2;
byte z = x + y; // Error, ya que la operación se hace como z = <int>x + <int>y
`````

#### ¡Cuidado con paréntesis!: debe abarcar toda la expresión:

`````java
byte x = 1;
byte y = 2;
byte z = (byte) x + y;    // equivalente a <byte>x + <int>y, promoción a <int>x + <int>y, luego resultado es <int>

System.out.println("z: " + z);
`````

#### Solución correcta:

`````java
byte x = 1;
byte y = 2;
byte z = (byte) (x + y); 

System.out.println("z: " + z);
`````

`````java
byte  x = 3;
float y = 1;
double z;
z = x + y; // x promociona a float y el resultado (float) promociona a double en la asignación

System.out.println("z: " + z);
`````

`````java
class Invocacion_Metodo {
    
    // con (a+b): (int)a + (double)b, a promociona de int a double -widening-: (double)a + b(double) -> (double)
    // con (byte)(a+b) el double (a+b) es convertido a byte con el casting (byte) -narrowing-
    // La función retorna un short, luego el (byte) que se retorna promociona a (short): -widening-
    static short funcion (int a, double b) {
        return (byte) (a+b);
    }
    
    public static void main(String[] args) {
        long x   = 1;
        byte p1  = 2;    /* CÁMBIESE posteriormente de tipo byte a tipo float y obsérvese consecuencias...*/
        float p2 = 3.4f;
        // promoción -widening- implícita del argumento "byte p1" al parámetro "int a", y del argumento "float p2" al parámetro "double b"
        x = funcion( p1, p2 );  
        System.out.println( "x = " + x );
    }
}

Invocacion_Metodo.main(null)
`````

**Literales enteros**: tipo forzado con *casting*

Si tipo entero del literal (`int` implícito o explícito, o `long` explícito) no es el tipo de la variable *destino*: *casting* obligatorio si hay pérdida de información y se desea **asumir**

`````java
byte x = 291;  // Error: 291 es un int y no cabe en rango de byte: [-128..127]
System.out.println("x: " + x);
`````

`````java
byte x = (byte) 291;
System.out.println("x: " + x);
// 291 en hexadecimal es 0x123, o en binario 0b0001_0010_0011
// Con casting a byte se elimina lo que esté por encima del byte: 
//    0x123 = 0x1_23 --se elimina el 1_--> 0x23 = 35
// o en binario
//    0b001_0010_0011 -> 0b0010_0011 = 35
// Se asume, por tanto, la pérdida de precisión (del 291 original al 35 defintivo)
`````

Con literales, no se sigue la regla de conversión implícita de operandos a tipo `int` (esta es la excepción pendiente de comentario anterior).

`````java
byte x = 1;
byte y = 2;
// byte z = x + y;  // Error ya que los operandos promocionan a int
byte z = (byte) (x + y); // Recuérdese que el casting era necesario ya que los operandos promocionan a `int`

System.out.println("z: " + z);

byte w = 1 + 2; // Con literales, no es necesario el casting, si el resultado cabe.
System.out.println("w: " + w);

//byte r = 234 + 76;  // El resultado no cabe!!! Error.
`````

- ### Variantes de `System.out. ...`

- `System.out.println(<VARIABLE>)`: imprime la `<VARIABLE>` y añade automáticamente una nueva línea
- `System.out.print(<VARIABLE>)`: idem que anterior, pero no añade una nueva línea
- `System.out.printf(<FORMATO>, <VARIABLE>)`: imprime la `<VARIABLE>` con el formato indicado en el, a su vez, string <FORMATO>. Numerosas Posiblidades para el string `<FORMATO>`:
    - `System.out.printf("%nd", <VARIABLE>);`: escribe `<VARIABLE>` (que debe ser de tipo numérico) en `n` espacios como un entero `d`ecimal
    - `System.out.printf("%-nd", <VARIABLE>);`: idem que anterior pero el número lo alinea a la izquierda
    - `System.out.printf("% nd", <VARIABLE>);`: idem que anterior pero en los `d` espacios deja uno para el signo (si es positivo aparece como un espacio en blanco)
    - `System.out.printf("%+nd", <VARIABLE>);`: idem que anterior pero si es positivo aparece con el signo `+`
    - `System.out.printf("%-+nd", <VARIABLE>);`: combinación de casos anteriores
    - `System.out.printf("%- nd", <VARIABLE>);`: combinación de casos anteriores

    - `System.out.printf("%ns", <VARIABLE>);`: escribe `<VARIABLE>` (que debe ser de tipo `String`) en `n` espacios como un string
    `System.out.printf("%-ns", <VARIABLE>);`: idem que anterior, pero alineado a la izquierda
    
    - `System.out.printf("%n.mf", <VARIABLE>);`: imprime la `<VARIABLE>` que debe ser real en `m` espacios (incluyendo signo y punto decimal); aparecerán `m` decimales.
        - Algunas de las variedades anteriores también pueden aplicar
    - `System.out.printf("%n.mE", <VARIABLE>);`: idem que anterior pero con notación científica o exponencial e incluyendo en los `m` espacios toda la información del esponente
    

`````java
String escala = "....................";
int x = -56;

System.out.println(escala);
System.out.println(x);
System.out.println();

System.out.println(escala);
System.out.printf("a:#%8d#", x);
System.out.printf("b:#%5d#\n", x);
System.out.println();


System.out.println(escala);
System.out.printf("c:#% 5d#\n", x);
System.out.printf("d:#% 5d#\n", -x);

System.out.println(escala);
System.out.printf("e:#%-5d#\n", x);
System.out.printf("f:#%-5d#\n", -x);


System.out.println(escala);
System.out.printf("g:#%+5d#\n", x);
System.out.printf("h:#%+5d#\n", -x);

System.out.println(escala);
System.out.printf("i:#%-+5d#\n", x);
System.out.printf("j:#%-+5d#\n", -x);


System.out.println(escala);
System.out.printf("k:#%- 5d#\n", x);
System.out.printf("l:#%- 5d#\n", -x);


System.out.println(escala);
System.out.printf("m:#% -5d#\n", x);
System.out.printf("n:#% -5d#\n", -x);


String saludo = "hola";
System.out.println(saludo);
System.out.printf("o:#%10s#\n", saludo);
System.out.printf("p:#%-10s\n#", saludo);



float z = -1234.567890f;
System.out.println(escala);
System.out.printf("q:#%+11.3f#\n", z);
System.out.printf("r:#%+11.3f#\n", -z);
System.out.printf("s:#% 11.3f#\n", z);
System.out.printf("t:#%11.3f#\n", z);
System.out.printf("u:#%11.3f#\n", -z);

System.out.printf("t:#%-11.3f#\n", z);
System.out.printf("u:#%-11.3f#\n", -z);
System.out.printf("v:#%-+11.3f#\n", -z);



System.out.printf("w:#%+11.3E#\n", -z);
`````

```java
<NOMBRE_DEL_METODO> ( [ARG_1 [, ARG_2 [...] ] ] );
```


- **Firma** de un método:
    - Nombre del método y el número y tipo de sus parámetros
    - El tipo de vuelta no interviene en la definición de _firma_

- Sutil distinción:
    - Los _parámetros_ de un método también se denominan _parámetros formales_
    - Los _parámetros_ con los que se invoca un método también se denominan _argumentos_ o _parámetros actuales_
    

`````java
int doblar(int x) {  // Parámetro: x
    int retorno;
    retorno = 2*x;
    return retorno;  // retorno del método
}


int z = 2;
int doble = doblar(z); // Argumento: z

System.out.printf( "Doble de %d: %d\n\n", z, doble );



// método void
void saludar(String nombre) {
    System.out.println("Hola " + nombre);
}

// Uso de return con métodos void
void saludar2(String nombre) {
    if (nombre.equals("Manolito")) return;
    System.out.println("Hola " + nombre);
}



saludar("Pepe");
saludar2("Manolito");
saludar2("Jaimito");

`````

### [2.4.2. Variables (locales)](#Índice)

- Variables declaradas, definidas y utilizadas en el contexto de un método
    - Cuando la variable lo está en el contexto de un objeto, se denominarán _atributos_

#### Ámbito de variables: ámbito de *bloque*

- Bloque de instrucciones: conjunto de instrucciones delimitadas por `{` y `}`

    ```java

    1 {
    2    int a;
    3    {
    4       int b;
    5       {
    6          ...
    7       }
    8    }
    9 }
    ```

    Tres bloques: 1--9, 3--8, 5--7

    `a` sólo es accesible desde su declaración hasta el cierre del bloque donde ha sido declarada: líneas 2-8

    `b` sólo es accesible desde su declaración hasta el cierre del bloque donde ha sido declarada: líneas 4-7


- Una variable no se puede *redeclarar* en el **mismo** ámbito, pero sí puede en ámbitos *disjuntos*:

`````java
boolean control = true;

if (control) {
    int x;
    x = 2;
    System.out.println("x: " + x);
} else {
    int x;
    x = 20;
    System.out.println("x: " + x);
    // int x = 100;               // Error por re-declaración en mismo bloque
    // Los dos siguientes errores no son emitidos por Jupyter Notebook debido a su funcionamiento
    //  boolean control = false;  // Error por re-declaración en el mismo bloque
    //  int control = 7;          // Error por re-declaración en el mismo bloque
}
`````

<div id="sobrecarga"></div>

### [2.4.4. Sobrecarga de métodos](#Índice)

- _Reutilización_ del **nombre** de un método en un nuevo método, pero con distinto **número o tipo** de parámetros
- Ejemplo:

`````java
int operar(int x, int y) {
    return x + y;
}

int operar(int x, int y, int z) {
    return x + y + z;
}

double operar(double x, double y) {
    return 10 * (x + y);
}


System.out.println(operar(1, 2));
System.out.println(operar(1, 2, 3));
System.out.println(operar(1., 2.));
`````

Cambiar **solo** el tipo de retorno no implicaría sobrecarga
- Sería una _redeclaración_ o _sobrescritura_ de un método, lo cual es un error (salvo en contextos de herencia, lo cual se estudiará más adelante)

`````java
int sumar(int x, int y) {
    return x + y;
}


// Esto sería un error de re-declaración
// En Jupyter Notebook se permite (sobrescritura) debido a la forma que tiene de emular Java

double sumar(int x, int y) {
    return (double) (10*x + y);
}

int x    = sumar(1,2);
double y = sumar(1, 2);
System.out.println(x);
System.out.println(y);
`````

<div id="parametros"></div>

### [2.4.5. Parámetros](#Índice)

- Los parámetros se pasan por valor
    - Se hace una copia del contenido del parámetro y esa es la información transferida al interior del método
    - Cualquier alteración dentro del parámetro del método no afecta al argumento

`````java
void metodo(int x) {
    x = 2*x;
    System.out.println("parámetro x: " + x);
}

int z = 2;
System.out.println("argumento z: " + z);
metodo(z);
System.out.println("argumento z: " + z);
`````

Para reflejar _fuera_ del ámbito del método la alteración de un parámetro:
- O se retorna el valor actualizado y se recoge fuera dicho valor
- O se modifica alguna propiedad de un objeto (se verá posteriormente)

`````java
class MiClase{
    int z;
}


int funcion(int x, MiClase objeto) {
    objeto.z = 4*x;
    return 2*x;
}

MiClase objeto = new MiClase();

int q = funcion(2, objeto);

System.out.println("objeto.z : " + objeto.z);
System.out.println("q: " + q);

`````

<div id="longvar"></div>

### [2.4.6. Longitud variable de parámetros](#Índice)

- Se pueden preparar los métodos para admitir su invocación con un número indeterminado de parámetros (del mismo tipo y al final...):
    - Notación `...`
    - La invocación se _camufla_ como si los parámetros hubieran sido codificados en un array en vez de en una lista.

`````java
void saludar(String saludo, String ... personas) {

    for (int i=0; i<personas.length; i++) {
        System.out.println(saludo + " " + personas[i]);
    }
    
    System.out.println();
}

saludar("Hola", "Pepito");
saludar("Hola", "Pepito", "Manolito");
saludar("Hola", "Pepito", "Manolito", "Jaimito");

// Alternativa
String[] amigos = {"Fulanito", "Menganito"};
saludar("¡Qué tal!", amigos);
`````

`````java
// Alternativa bucle for 

void saludar(String saludo, String ... personas) {

    for (String persona: personas) {
        System.out.println(saludo + " " + persona);
    }
    
    System.out.println();
}

saludar("Hola", "Pepito");
saludar("Hola", "Pepito", "Manolito");
saludar("Hola", "Pepito", "Manolito", "Jaimito");

// Alternativa
String[] amigos = {"Fulanito", "Menganito"};
saludar("¡Qué tal!", amigos);
`````
